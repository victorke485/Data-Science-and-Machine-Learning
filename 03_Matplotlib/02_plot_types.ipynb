```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# line plot
x = np.linspace(0, 100, 10)
y = np.square(x)
fig, ax = plt.subplots()
ax.plot(x, y)
```

```python
# scatter plot
fig, ax = plt.subplots()
ax.scatter(x, y)
```

```python
# Plot from a dictionary
language_popurality = {
    "C": 6,
    "Python": 10,
    "Javascript": 9,
    "Lua": 3,
    "C++": 7,
    "HolyC": 2,
}
fig, ax = plt.subplots()
ax.bar(language_popurality.keys(), language_popurality.values())
ax.set(
    title = "Programming Languages Popularity",
    xlabel = "Language",
    ylabel = "Popularity"
)
```

```python
# Horizontal bar
fig, ax = plt.subplots()
ax.barh(language_popurality.keys(), language_popurality.values())
ax.set(
    title = "Programming Languages Popularity",
    xlabel = "Popularity",
    ylabel = "Language"
)
```

```python
# Histogram
x = np.random.randn(1000)
fig, ax = plt.subplots()
ax.hist(x)
```

```python
# Subplots - option 1
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(10, 5))

x = np.linspace(0, 100, 10)

fig.suptitle("Subplots")

ax1.plot(x, x/2)
ax2.scatter(x, x**2)
ax3.bar(language_popurality.keys(), language_popurality.values())
ax4.hist(np.random.randn(1000))
```

```python
# Subplots - option2
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10, 5))
ax[0, 0].plot(x, x/2)
ax[0, 1].scatter(x, x**2)
ax[1, 0].bar(language_popurality.keys(), language_popurality.values())
ax[1, 1].hist(np.random.randn(1000))
```
