```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
car_sales = pd.read_csv("car-sales.csv")
car_sales.head(10)
```

```python
car_sales.plot()
```

```python
car_sales["Price"] = car_sales["Price"].str.replace(pat={"$": "", ",": "", ".": ""}).astype(int) / 100
car_sales["Price"]
```

```python
car_sales["Sale Date"] = pd.date_range(start="1/1/2025", periods=len(car_sales))
car_sales
```

```python
car_sales["Price"] = car_sales["Price"].astype(int)
car_sales["Total Sales"] = car_sales["Price"].cumsum()
car_sales
```

```python
# Total sales plot
car_sales.plot(x="Sale Date", y="Total Sales")
```

```python
car_sales.plot(x="Odometer (KM)", y="Price", kind="scatter")
```
