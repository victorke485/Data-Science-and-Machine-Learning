```python
import pandas as pd
```

```python
car_sales = pd.read_csv("car-sales-missing-data.csv")
car_sales.head(10)
```

```python
car_sales["Odometer"] = car_sales["Odometer"].fillna(car_sales["Odometer"].mean())
# car_sales["Odometer"].fillna(car_sales["Odometer"].mean(), inplace=True)
car_sales.dropna(inplace=True)

car_sales
```

```python
# Column from series
seat_column = pd.Series([5,5,5])

# New column called seats
car_sales["Seats"] = seat_column
car_sales
```

```python
car_sales["Seats"] = car_sales["Seats"].fillna(5)
car_sales
```

```python
# Column from Python list - has to be the same length as the dataframe
fuel_economy = [7.5, 9.2, 5.0, 9.3, 8.7, 6.7]
car_sales["Fuel per 100KM"] = fuel_economy
car_sales
```

```python
# Column from other columns
car_sales["Total fuel used"] = car_sales["Odometer"] / 100 * car_sales["Fuel per 100KM"] 
car_sales
```

```python
# Creating a column fom a single value
car_sales["Number of Wheels"] = 4
car_sales
```

```python
car_sales["Passed road safety"] = True
car_sales
```

```python
# Dropping column
car_sales = car_sales.drop("Total fuel used", axis=1)
car_sales
```

```python
car_sales = car_sales.drop(5, axis=0)
car_sales
```
