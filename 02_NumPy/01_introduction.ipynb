```python
# Stands for numerial Python
# Used to do large scale numerial calculation because it is fast
# Is fast
# Behind the scenes optimizations written in C
# Vectorization via broadcasting (avoiding loops)
```

```python
import numpy as np
```

```python
# DataTypes & Attributes

# NumPy's main datatype is ndarray

a1 = np.array([1, 2, 3, 4])
a2 = np.array([[1, 2, 3, 4.8], [4, 5, 6.7, 9]])
a3 = np.array([ [[2.2, 3, 8], [54, 34, 2.2], [12, 23, 3.6]], [[43.4, 1.1, 4.5], [5, 8.9, 12], [123, 32.23, 10]] ])

print(a1, "\n\n")
print(a2, "\n\n")
print(a3)
```

```python

# shape
print(a1.shape)
print(a2.shape)
print(a3.shape)
```

```python
# Number of dimensions
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
```

```python
# dtype
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

```python
# size
print(a1.size)
print(a2.size)
print(a3.size)
```

```python
# type
print(type(a1))
print(type(a2))
print(type(a3))
```
