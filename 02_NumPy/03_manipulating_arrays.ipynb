```python
import numpy as np
a1 = np.array([1, 2, 3, 4, 5])
a2 = np.array([ [3, 4, 5, 6, 7], [6, 7, 8, 9, 2] ])
ones = np.ones(5)
print(a1)
print(a2)
print(ones)
```

```python
# Arithmetic
print(a1 + ones)
print(a1 - ones)
print(a1 * ones)
print(a1 * a2)
print(a1 / a2)
print(a1 // a2)
print(a1 ** a2)

print("------------")
print(np.square(a1))
print(np.add(a1, ones))
print(np.mod(a1, ones))
```

```python
# Aggregation - performing the same operation on a number of things
my_array = np.array([5, 43, 23, 12, 21, 23])

print(sum(my_array))
print(np.sum(my_array))

# NB: Use Python's methods on Python datatypes and use NumPy's methods on NumPy arrays
```

```python
# Creating a massive NumPy array
massive_array = np.random.random(100000)
massive_array[:10]
```

```python
%timeit (sum(massive_array))
%timeit np.sum(massive_array)
```

```python
print(np.mean(massive_array))
print(np.max(massive_array))
print(np.min(massive_array))
print(np.std(massive_array))
# Standard deviation - how spread out a group of numbers is from the mean (square root of variance)
print(np.sqrt(np.var(massive_array)))
print(np.var(massive_array))
# Variance is the measure of the average degree to which each number is different to the mean
```

```python
# Reshaping & Transposing
array1 = np.array([[2, 3, 4], [6, 7, 8]])
array2 = np.array([ [[5, 6, 7], [12, 11, 32], [12, 11, 54]], [[23, 32, 12], [12, 43, 54], [67, 65, 43]] ])

array1.shape, array2.shape
```

```python
array1_reshape = array1.reshape(2, 3, 1)
print(array1_reshape, array1_reshape.shape)
array1_reshape * array2
```

```python
# Transpose - switches the axis
print(array1)
array1.T
```

```python
# Dot product
np.random.seed(0)
mat1 = np.random.randint(10, size=(5,3))
mat2 = np.random.randint(10, size=(5,3))
mat1, mat2
```

```python
mat1 * mat2
```

```python
mat3 = np.dot(mat1, mat2.T)
mat3, mat3.shape
```

```python
# Comparison Operators
mat1 > mat2
mat1 >= mat2
mat1 == mat2
mat1 <= mat2
mat1 != mat2
mat1 < mat2
```
