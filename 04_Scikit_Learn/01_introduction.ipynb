**Scikit-Learn (sklearn)** is a python machine learning framework.

Why Scikit-Learn:
- Built on Numpy and Matplotlib (and python)
- Has many in-built machine learning models
- Methods to evaluate your machine learning models
- Very well-designed API


Scikit-Learn workflow:
1. Get data ready
2. Pick a model
3. Fit the model to the data and make a prediction
4. Evaluate the model
5. Improve through exprementation
6. Save and reload your trained model

```python
# An end-to-end Scikit-Learn Workflow
# Get data ready

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import pickle


heart_disease = pd.read_csv("heart-disease.csv")
heart_disease
```

```python
# Create X  (feature matrix)
x = heart_disease.drop("target", axis=1)

# Create y (labels)
y = heart_disease["target"]
```

```python
# Choose the right model and hyperparameters
clf = RandomForestClassifier(n_estimators=100)

clf.get_params(deep=True)
```

```python
# Fit the model to the training data

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
clf.fit(x_train, y_train)
```

```python
# Make a prediction
y_preds = clf.predict(x_test)
y_preds
```

```python
# Evaluate the model on the training data and test data
clf.score(x_train, y_train)
```

```python
clf.score(x_test, y_test)
```

```python

print(classification_report(y_test, y_preds))
```

```python
confusion_matrix(y_test, y_preds)
```

```python
accuracy_score(y_test, y_preds)
```

```python
# Improve a model
# Try different amount of n_estimators
best_score = 0
best_clf = None
for i in range(10, 110, 10):
    clf = RandomForestClassifier(n_estimators=i)
    clf.fit(x_train, y_train)
    y_preds = clf.predict(x_test)
    score = accuracy_score(y_test, y_preds) * 100
    if score > best_score:
        best_score = score
        best_clf = clf
    print(f"Estimator: {i}, accuracy score: {score:.2f}%")

print(f"Best score: {best_score}")
```

```python
# Save model and load it
with open("random_forest_model_1.pkl", "wb") as file:
    pickle.dump(best_clf, file)
```
