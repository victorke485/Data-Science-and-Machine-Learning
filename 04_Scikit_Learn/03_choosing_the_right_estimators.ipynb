```python
import pandas as pd
import numpy as np

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
```

```python
# Get Californina Housing dataset
housing = fetch_california_housing()
housing
```

```python
housing_df = pd.DataFrame(housing["data"], columns=housing["feature_names"])
housing_df["target"] = housing["target"]
housing_df
# housing_df.isna().sum()
```

```python
# Split data
x = housing_df.drop(columns=["target"])
y = housing_df["target"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Select and train model
# RidgeRegression
model = Ridge()
model.fit(x_train, y_train)
print(f"Ridge score: {model.score(x_test, y_test)}")

# SVR
model = SVR()
model.fit(x_train, y_train)
print(f"SVR score: {model.score(x_test, y_test)}")

# Ensemble - combination of smaller models to try and make better predictions
# RandomForest
model = RandomForestRegressor()
model.fit(x_train, y_train)
print(f"RandomForestRegressor score: {model.score(x_test, y_test)}")
```
