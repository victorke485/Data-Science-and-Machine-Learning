```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# -----------------------------
# Experimental data
# -----------------------------

mass = np.array([50,100,150,200,250,300])       # grams
load = mass/1000 * 9.81                         # Newtons

quarter = np.array([0.07,0.06,0.15,0.29,0.32,0.40])
half = np.array([0.00,0.07,0.125,0.31,0.475,0.63])

# -----------------------------
# Plot
# -----------------------------

plt.figure(figsize=(8,6))

plt.plot(
    quarter,
    load,
    'o-',
    linewidth=2,
    label='Quarter Span'
)

plt.plot(
    half,
    load,
    's-',
    linewidth=2,
    label='Half Span'
)

plt.xlabel("Deflection")
plt.ylabel("Load (N)")
plt.title("Load versus Deflection for a Simply Supported Beam")

plt.legend()

plt.tight_layout()

plt.show()
```
